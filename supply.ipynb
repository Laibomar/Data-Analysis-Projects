DATA ANALYSIS

```python
#Total revenue
total_revenue = data['Sales'].sum()

print(f"Total revenue is : {total_revenue}$")
```

Products analysis

```python
#The categories with the most benefit
most_revenue_cat = data.groupby('Category')['Sales'].sum().reset_index()

most_revenue_cat = most_revenue_cat.sort_values(by='Sales', ascending=False)

print(most_revenue_cat)
```

```python
fig1 = px.pie(data_frame=most_revenue_cat, names='Category', values='Sales')
fig1.update_layout(title="Revenue distribution by categoy")
fig1.show()
```

```python
# Categories sales
most_sal_cat = data.groupby('Category')['Quantity'].sum().reset_index()

most_sal_cat = most_sal_cat.sort_values(by='Quantity', ascending=False)

print(most_sal_cat)
```

```python
fig1 = px.pie(data_frame=most_sal_cat, names='Category', values='Quantity')
fig1.update_layout(title="Categories sales")
fig1.show()
```

```python
# Top 10 subcategories with the most benefit
most_revenu_subcat = data.groupby('Sub-Category')['Sales'].sum().reset_index()

most_revenu_subcat = most_revenu_subcat.sort_values(by='Sales', ascending=False)

most_revenu_subcat = most_revenu_subcat.head(10)

print(most_revenu_subcat)
```

```python
fig2 = px.bar(data_frame=most_revenu_subcat, x='Sub-Category', y='Sales')
fig2.update_layout(title="Top 10 subcategories revenue")
fig2.show()
```

```python
# Top 10 sales subcategories
most_sales_subcat = data.groupby('Sub-Category')['Quantity'].sum().reset_index()

most_sales_subcat = most_sales_subcat.sort_values(by='Quantity', ascending=False)

most_sales_subcat = most_sales_subcat.head(10)

print(most_sales_subcat)
```

```python
fig2 = px.bar(data_frame=most_sales_subcat, x='Sub-Category', y='Quantity')
fig2.update_layout(title="Top 10 sales subcategories")
fig2.show()
```

Geographic analysis

```python
#By region
reg_sales = data.groupby('Region')['Sales'].sum().reset_index()

reg_sales = reg_sales.sort_values(by='Sales', ascending=False)

print(reg_sales)
```

```python
fig3 = px.pie(data_frame=reg_sales, names='Region', values='Sales')
fig3.update_layout(title="Revenue distribution By Region")
fig3.show()
```

```python
# By state
state_sales = data.groupby('State')['Sales'].sum().reset_index()

state_sales = state_sales.sort_values(by='Sales', ascending=False)

state_sales = state_sales.head(10)

print(state_sales)
```

```python
fig4 = px.bar(data_frame=state_sales, x='State', y='Sales')
fig4.update_layout(title="Top 10 states revenue")
fig4.show()
```

Segment analysis

```python
# Revenue by segment
seg_sales = data.groupby('Segment')['Sales'].sum().reset_index()

seg_sales = seg_sales.sort_values(by='Sales', ascending=False)

print(seg_sales)
```

```python
fig5 = px.pie(data_frame=seg_sales, names='Segment', values='Sales', color='Segment',
              color_discrete_map={'Home Office':'lightcyan',
                                 'Corporate':'cyan',
                                 'Consumer':'royalblue',
                                 }, 
              title="Revenue distribution by segment",)
fig5.show()
```

```python
data.columns
```

```python
# Converting "Date" colums to the right format

data['Order Date'] = pd.to_datetime(data['Order Date'], format="%Y-%m-%d")

data['Ship Date'] = pd.to_datetime(data['Ship Date'], format="%Y-%m-%d")
```

```python
data['Month'] = data['Ship Date'].dt.month

data['Month'] = data['Month']

# Calculation of the monthly sales trend
monthly_sales = data.groupby('Month')['Sales'].sum().reset_index()
```

```python
fig6 = px.line(data_frame=monthly_sales, x='Month', y='Sales', markers=True, title='Monthly sales')
fig6.show()
```

```python
# Add column year-mounth
data['Year-Month'] = data['Ship Date'].dt.to_period('M')


data['Year-Month'] = data['Year-Month'].astype(str)

# Sales by mounth/year
monthly_sales_by_year = data.groupby('Year-Month')['Sales'].sum().reset_index()
```

```python
fig6 = px.line(data_frame=monthly_sales_by_year, x='Year-Month', y='Sales')
fig6.show()
```

Returns analysis

```python
returns = data[data['Returned'] == 'Yes']

returns_by_category = returns.groupby('Category')['Sales'].sum().reset_index()
```

```python
returns_by_category
```

```python
fig8 = px.bar(data_frame=returns_by_category, x='Category', y='Sales',
              title='Returned sales by categories')
fig8.show()
```

```python
returns_by_subcategory = returns.groupby('Sub-Category')['Sales'].sum().reset_index()

returns_by_subcategory = returns_by_subcategory.sort_values(by='Sales', ascending=False)
```

```python
fig8 = px.bar(data_frame=returns_by_subcategory, x='Sub-Category', y='Sales', color='Sales',
              text_auto='.2s',
              title='Returned sales by subcategories')
fig8.show()
```

```python
# Analysis of sales with and without discounts
pos_discount = data[data['Discount'] > 0]

sales_with_discount = pos_discount.groupby('Discount')['Sales'].sum().reset_index()
```

```python
# Discount visualisation
fig8 =  px.line(data_frame=sales_with_discount, x='Discount', y='Sales', text='Discount', 
                title="Sales distribution by discounts")
fig8.update_traces(textposition="bottom right", line = dict(color='blue'))
fig8.show()
```
